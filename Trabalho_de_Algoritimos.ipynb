```python
def calcular_valor_liquido_impostos(valor_bruto, aliquota_icms, aliquota_pis, aliquota_cofins):
    """
    Função para calcular o valor líquido após descontos de ICMS, PIS e COFINS.

    :param valor_bruto: valor bruto da venda ou faturamento
    :param aliquota_icms: alíquota do ICMS (em percentual)
    :param aliquota_pis: alíquota do PIS (em percentual)
    :param aliquota_cofins: alíquota do COFINS (em percentual)
    :return: valor líquido após descontos
    """
    # Calculando os valores dos impostos
    desconto_icms = valor_bruto * (aliquota_icms / 100)
    desconto_pis = valor_bruto * (aliquota_pis / 100)
    desconto_cofins = valor_bruto * (aliquota_cofins / 100)

    # Calculando o valor líquido
    valor_liquido = valor_bruto - desconto_icms - desconto_pis - desconto_cofins

    return valor_liquido

# Entrada de dados interativa
valor_bruto = float(input("Digite o valor bruto: R$ "))  # valor bruto da venda ou faturamento
aliquota_icms = float(input("Digite a alíquota do ICMS (%): "))  # alíquota do ICMS
aliquota_pis = float(input("Digite a alíquota do PIS (%): "))  # alíquota do PIS
aliquota_cofins = float(input("Digite a alíquota do COFINS (%): "))  # alíquota do COFINS

# Calculando o valor líquido
valor_liquido = calcular_valor_liquido_impostos(valor_bruto, aliquota_icms, aliquota_pis, aliquota_cofins)

# Exibindo o resultado
print(f"\nValor bruto: R${valor_bruto:.2f}")
print(f"Valor líquido após os descontos de ICMS, PIS e COFINS: R${valor_liquido:.2f}")
```
